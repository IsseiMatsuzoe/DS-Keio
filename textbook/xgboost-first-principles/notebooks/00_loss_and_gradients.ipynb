# 00 — 損失関数・微分・最適化
**目標:** MSE と予測値に関する勾配を式から導き、数値微分で検証する。

教師あり学習では入力 $x_i$ に対し予測 $\hat y_i=f(x_i;\theta)$ を出す。回帰の最も基本的な損失は

$$L(\hat{\boldsymbol y})=\frac1n\sum_{i=1}^n(\hat y_i-y_i)^2.$$

各 $\hat y_i$ に関して偏微分すると

$$\frac{\partial L}{\partial \hat y_i}=\frac2n(\hat y_i-y_i).$$

注意：後の Boosting では $\frac12\sum_i(\hat y_i-y_i)^2$ を使い、負の勾配が **残差** $y_i-\hat y_i$ に一致するようにする。定数倍は実装上の学習率へ吸収できるが、数式の定義を混同しないこと。

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-2, 2, 101)
y = x**2 + 0.3*x
plt.plot(x, y)
plt.scatter([0], [0], label="minimum vicinity")
plt.title("Loss landscape: curvature and slope")
plt.xlabel("parameter"); plt.ylabel("loss"); plt.legend(); plt.show()
```

## 手計算から有限差分へ
中心差分 $\partial L/\partial p_i\approx[L(p+\epsilon e_i)-L(p-\epsilon e_i)]/(2\epsilon)$。

1. 例: $y=(1,2)$, $\hat y=(3,4)$ の MSE はいくつ？
2. そのとき $\nabla_{\hat y}L$ は？
3. $\epsilon$ を大きくしすぎる／小さくしすぎると何が起きる？

```python
y = np.array([1., 2.])
p = np.array([3., 4.])
mse_demo = np.mean((p-y)**2)
analytical = 2*(p-y)/len(y)
epsilon = 1e-5
numerical = np.array([
    (np.mean((p+epsilon*np.eye(2)[j]-y)**2)
    -np.mean((p-epsilon*np.eye(2)[j]-y)**2))/(2*epsilon)
    for j in range(2)
])
print("MSE:", mse_demo)
print("analytic:", analytical, "finite difference:", numerical)
assert np.allclose(analytical, numerical)
```
