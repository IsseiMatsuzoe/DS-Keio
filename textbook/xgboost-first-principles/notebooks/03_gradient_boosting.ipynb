# 03 — Gradient Boosting：勾配を関数で近似する
## モデルを足していく
$$F_M(x)=F_0(x)+\eta\sum_{m=1}^M h_m(x),\qquad F_0=\bar y.$$

通常の二乗誤差では $L=\frac12\sum_i(y_i-F(x_i))^2$ とすると

$$-\frac{\partial L}{\partial F(x_i)} = y_i-F(x_i)\equiv r_i.$$

つまり **現状の残差 $r_i$ に新しい木 $h_m$ を fit** し、学習率 $\eta$ で足し込むのが最も素朴な Gradient Boosting。これは厳密な意味では木のパラメータを微分する手法ではなく、**予測値（関数値）空間の負の勾配** を木で近似する方法である。

### 手順
1. $F_0=\mathrm{mean}(y)$
2. $r^{(m)}=y-F_{m-1}(x)$
3. $h_m=\mathrm{fit\_stump}(x,r^{(m)})$
4. $F_m=F_{m-1}+\eta h_m$

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingRegressor
x = np.linspace(0, 1, 100)
y = np.where(x < .5, -1, 2) + .15*np.sin(8*np.pi*x)
stages = [1, 2, 5, 15, 60]
fig, ax = plt.subplots(figsize=(9,4))
ax.scatter(x,y,s=8,label="target")
for n in stages:
    m = GradientBoostingRegressor(max_depth=1,n_estimators=n,learning_rate=.1,random_state=0)
    m.fit(x[:,None],y)
    ax.plot(x,m.predict(x[:,None]),label=f"{n} stages")
ax.set_title("Successive corrections from shallow trees")
ax.legend(); plt.show()
```
